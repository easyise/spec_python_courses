
# Обработка видео и последовательностей в Python

В этом ноутбуке мы рассмотрим основные методы обработки видео и последовательностей кадров с использованием библиотек OpenCV и других инструментов. 
Темы включают:
- Загрузка и отображение видео
- Извлечение кадров
- Обработка последовательностей кадров
- Сохранение видео


```

import cv2
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Video, display
```


## Загрузка и отображение видео

Для начала загрузим видеофайл и отобразим его.


```

# Путь к видеофайлу (замените на путь к вашему видео)
video_path = 'sample_video.mp4'

# Отображение видео в Jupyter (если поддерживается)
display(Video(video_path, embed=True))
```


## Извлечение кадров из видео

Извлечем кадры из видео с помощью OpenCV.


```

cap = cv2.VideoCapture(video_path)
frames = []
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    frames.append(frame)
cap.release()

print(f"Извлечено {len(frames)} кадров")
```


## Отображение нескольких кадров

Посмотрим на первые несколько кадров видео.


```

for i in range(3):
    frame_rgb = cv2.cvtColor(frames[i], cv2.COLOR_BGR2RGB)
    plt.imshow(frame_rgb)
    plt.title(f'Кадр {i+1}')
    plt.axis('off')
    plt.show()
```


## Обработка последовательностей кадров

Применим простую фильтрацию: преобразование кадров в оттенки серого.


```

gray_frames = [cv2.cvtColor(f, cv2.COLOR_BGR2GRAY) for f in frames]

# Отобразим первый обработанный кадр
plt.imshow(gray_frames[0], cmap='gray')
plt.title('Первый кадр (оттенки серого)')
plt.axis('off')
plt.show()
```


## Сохранение обработанного видео

Сохраним видео после обработки.


```

out_path = 'output_video.avi'
height, width = gray_frames[0].shape
out = cv2.VideoWriter(out_path, cv2.VideoWriter_fourcc(*'XVID'), 20.0, (width, height), isColor=False)

for frame in gray_frames:
    out.write(cv2.cvtColor(frame, cv2.COLOR_GRAY2BGR))
out.release()

print("Видео сохранено как", out_path)
```
