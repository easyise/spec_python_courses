$E = \sqrt{\sum_{i=1}^{D}\frac{(y_i-y\_pred_i)^2)}{ n }}$

$E = \sqrt{\sum_{i=1}^{D}\frac{(y_i-(k*x_i+b))^2)}{ n }}$

частичная производная по k, в одномерном случае

$\frac{\partial E}{\partial k} = -(y - y\_pred) * x $

частичная производная по b, в одномерном случае

$\frac{\partial E}{\partial k} = -(y - y\_pred) $

### Accuracy (точность)

$Accuracy = \frac{TP+TN}{TP+TN+FP+FN}$

### ROC-AUC

$Precision = \frac{TP}{TP+FP}$
$Recall = \frac{TP}{TP+FN}$

$TPrate = Recall = \frac{TP}{TP+FN}$

$FPrate = \frac{FP}{FP+TN}$


### ReLU

Rectfied Linear Unit - "ограниченная линейная функция"

$Relu(z) = max(0, z)$

### Sigmoid

Или "логистическая сигмоида".

$\sigma(z) = \frac{1} {1 + e^{-z}}$

### Tanh

Гиперболический тангенс. 

$tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}} = \frac{1 - e^{-2x}}{1 + e^{-2x}}$

### Softmax

Используется в задачах $N$-классовой классификации.

$\sigma(z_i) = \frac{e^{z_{i}}}{\sum_{j=1}^K e^{z_{j}}} \ \ \ for\ i=1,2,\dots,K$
